```python
# READ parquet WITH DUCKDB
pqt_str =pqt_file.as_posix()

conn = duckdb.connect()
query = f"create table if not exists aggtrades_v1 as select * from read_parquet('{pqt_str}') "

conn.sql(query=query)
conn.sql("select * from aggtrades_v1 ")
```
